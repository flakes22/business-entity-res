# Step 2: Normalization
Normalize business names and addresses: lowercase, unicode normalization, abbreviation expansion, punctuation removal, and legal suffix stripping.

```python
import re
import unicodedata

# Output directory for normalized files
NORMALIZED_DIR = "/content/drive/MyDrive/amazon_ml/6ab10eb3b23ba_student_resource/student_resource/dataset/normalized/"
os.makedirs(NORMALIZED_DIR, exist_ok=True)
```

```python
# Standardize abbreviations
NAME_REPLACEMENTS = {
    r'\binc\b': 'incorporated',
    r'\bcorp\b': 'corporation',
    r'\bltd\b': 'limited',
    r'\bpvt\b': 'private',
    r'\bllc\b': 'limited liability company',
    r'\bco\b': 'company'
}

ADDRESS_REPLACEMENTS = {
    r'\bst\b': 'street',
    r'\brd\b': 'road',
    r'\bave\b': 'avenue',
    r'\bblvd\b': 'boulevard',
    r'\bdr\b': 'drive',
    r'\bln\b': 'lane',
    r'\bct\b': 'court',
    r'\bapt\b': 'apartment',
    r'\bste\b': 'suite',
    r'\bhwy\b': 'highway'
}

LEGAL_SUFFIXES_TO_REMOVE = r'\b(incorporated|corporation|limited|private|limited liability company|company)\b'
```

```python
def clean_text(text, replacements):
    """
    Lowercases, normalizes unicode, handles '&', and standardizes.
    """
    if pd.isna(text):
        return ""
    
    # Lowercase
    text = str(text).lower()
    
    # Unicode normalization (e.g., handles French accents well without destroying Hindi characters)
    text = unicodedata.normalize('NFKC', text)
    
    # Explicitly handle '&' vs 'and' before removing punctuation
    text = text.replace('&', ' and ')
    
    # Apply standard abbreviation replacements
    for pattern, replacement in replacements.items():
        text = re.sub(pattern, replacement, text)
        
    # Remove punctuation except spaces (keep unicode characters for Hindi/French)
    text = re.sub(r'[^\w\s]', ' ', text)
    
    # Remove multiple spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

def remove_legal_suffixes(text):
    """Removes expanded legal suffixes from the normalized text."""
    text = re.sub(LEGAL_SUFFIXES_TO_REMOVE, '', text)
    return re.sub(r'\s+', ' ', text).strip()

def normalize_dataframe(df):
    """
    Normalizes the names and addresses in the dataframe, maintaining original representations.
    """
    print(f"Normalizing dataframe with {len(df)} rows...")
    
    # Handle missing values
    df['business_name'] = df['business_name'].fillna("")
    df['business_address'] = df['business_address'].fillna("")
    
    # Clean Text (creates new columns)
    print("Cleaning business names...")
    df['norm_name'] = df['business_name'].apply(lambda x: clean_text(x, NAME_REPLACEMENTS))
    
    print("Creating name without legal suffixes...")
    df['norm_name_no_suffix'] = df['norm_name'].apply(remove_legal_suffixes)
    
    print("Cleaning business addresses...")
    df['norm_address'] = df['business_address'].apply(lambda x: clean_text(x, ADDRESS_REPLACEMENTS))
    
    return df
```

```python
def process_and_save(filename, source_dir):
    """Loads, normalizes, and saves the dataframe."""
    print(f"\n--- Processing {filename} ---")
    out_filepath = os.path.join(NORMALIZED_DIR, filename)
    if os.path.exists(out_filepath):
        print(f"File {out_filepath} already exists. Skipping.")
        return
        
    filepath = os.path.join(source_dir, filename)
    
    # Load dataset
    df = pd.read_csv(filepath, sep="\t")
    
    # Normalize
    df_norm = normalize_dataframe(df)
    
    # Save to output directory
    df_norm.to_csv(out_filepath, sep="\t", index=False)
    print(f"Saved normalized data to {out_filepath}")
```

```python
print("Starting Normalization Process...")

# Train files
train_files = ["train_source1.tsv", "train_source2.tsv", "train_source3.tsv"]
for file in train_files:
    process_and_save(file, TRAIN_DIR)
    
# Test files
test_files = ["test_source1.tsv", "test_source2.tsv", "test_source3.tsv"]
for file in test_files:
    process_and_save(file, TEST_DIR)

print("\nNormalization Complete! All files saved in", NORMALIZED_DIR)
```

# Step 3: Blocking
Generate candidate pairs using multiple blocking strategies: name prefix, token overlap, address tokens, and TF-IDF (char n-grams).

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors
import gc
from collections import defaultdict

# Paths for blocking
NORM_DIR = "/content/drive/MyDrive/amazon_ml/6ab10eb3b23ba_student_resource/student_resource/dataset/normalized/"
GT_DIR = "/content/drive/MyDrive/amazon_ml/6ab10eb3b23ba_student_resource/student_resource/dataset/train/"
CANDIDATES_DIR = "/content/drive/MyDrive/amazon_ml/6ab10eb3b23ba_student_resource/student_resource/dataset/candidates/"

os.makedirs(CANDIDATES_DIR, exist_ok=True)
```

### Block 1: Name Prefix Blocking (Country + first N chars)

```python
def block_name_prefix(s1, s_target, prefix_lengths=[3, 4, 5]):
    """Multiple name prefix blocks to catch variations."""
    all_pairs = []
    for n in prefix_lengths:
        s1_key = s1['country'].astype(str) + "_" + s1['norm_name_no_suffix'].astype(str).str[:n]
        t_key = s_target['country'].astype(str) + "_" + s_target['norm_name_no_suffix'].astype(str).str[:n]
        
        s1_temp = pd.DataFrame({'entity_id_s1': s1['entity_id'], 'bk': s1_key})
        t_temp = pd.DataFrame({'entity_id_cand': s_target['entity_id'], 'bk': t_key})
        
        # Filter out very short/empty keys
        s1_temp = s1_temp[s1_temp['bk'].str.len() > (3 + n)]
        t_temp = t_temp[t_temp['bk'].str.len() > (3 + n)]
        
        # Cap block size to avoid explosion from common prefixes
        t_counts = t_temp['bk'].value_counts()
        valid_keys = t_counts[t_counts <= 500].index
        t_temp = t_temp[t_temp['bk'].isin(valid_keys)]
        
        merged = pd.merge(s1_temp, t_temp, on='bk')[['entity_id_s1', 'entity_id_cand']]
        all_pairs.append(merged)
    
    return pd.concat(all_pairs).drop_duplicates() if all_pairs else pd.DataFrame(columns=['entity_id_s1', 'entity_id_cand'])
```

### Block 2: Token Overlap Blocking (shared rare name words)

```python
def block_name_tokens(s1, s_target, max_token_freq=200):
    """Block on shared rare/informative name tokens within same country."""
    # Build token -> entity_id index for target
    target_index = defaultdict(list)
    for _, row in s_target.iterrows():
        country = str(row['country'])
        name = str(row['norm_name_no_suffix'])
        tokens = name.split()
        for tok in tokens:
            if len(tok) >= 4:  # Only use tokens >= 4 chars (skip 'the', 'and', etc.)
                key = country + "_" + tok
                target_index[key].append(row['entity_id'])
    
    # Cap: remove overly common tokens
    target_index = {k: v for k, v in target_index.items() if len(v) <= max_token_freq}
    
    pairs = []
    for _, row in s1.iterrows():
        country = str(row['country'])
        name = str(row['norm_name_no_suffix'])
        tokens = name.split()
        cands = set()
        for tok in tokens:
            if len(tok) >= 4:
                key = country + "_" + tok
                if key in target_index:
                    cands.update(target_index[key])
        for c in cands:
            pairs.append({'entity_id_s1': row['entity_id'], 'entity_id_cand': c})
    
    return pd.DataFrame(pairs).drop_duplicates() if pairs else pd.DataFrame(columns=['entity_id_s1', 'entity_id_cand'])
```

### Block 3: Address Token Blocking (shared address components)

```python
def block_address_tokens(s1, s_target, max_token_freq=300):
    """Block on shared rare address tokens (street numbers, city names, PINs)."""
    # Extract significant address tokens: numbers, long words
    def get_addr_tokens(addr):
        addr = str(addr)
        tokens = addr.split()
        # Keep tokens that are: numbers, or words >= 5 chars
        return [t for t in tokens if (t.isdigit() and len(t) >= 3) or (len(t) >= 5 and t.isalpha())]
    
    # Build index for target
    target_index = defaultdict(list)
    for _, row in s_target.iterrows():
        country = str(row['country'])
        tokens = get_addr_tokens(row['norm_address'])
        for tok in tokens:
            key = country + "_" + tok
            target_index[key].append(row['entity_id'])
    
    target_index = {k: v for k, v in target_index.items() if len(v) <= max_token_freq}
    
    pairs = []
    for _, row in s1.iterrows():
        country = str(row['country'])
        tokens = get_addr_tokens(row['norm_address'])
        cands = set()
        for tok in tokens:
            key = country + "_" + tok
            if key in target_index:
                cands.update(target_index[key])
        for c in cands:
            pairs.append({'entity_id_s1': row['entity_id'], 'entity_id_cand': c})
    
    return pd.DataFrame(pairs).drop_duplicates() if pairs else pd.DataFrame(columns=['entity_id_s1', 'entity_id_cand'])
```

### Block 4: TF-IDF Blocking (char n-grams, per country)

```python
def block_tfidf(s1, s_target, text_column, top_k=10, batch_size=10000, distance_threshold=0.7):
    """TF-IDF char n-gram blocking, done PER COUNTRY to enforce country match."""
    countries = set(s1['country'].dropna().unique()) & set(s_target['country'].dropna().unique())
    
    all_pairs = []
    for country in countries:
        print(f"    TF-IDF [{text_column}] for country: {country}")
        s1_c = s1[s1['country'] == country].reset_index(drop=True)
        t_c = s_target[s_target['country'] == country].reset_index(drop=True)
        
        if len(s1_c) == 0 or len(t_c) == 0:
            continue
        
        s1_texts = s1_c[text_column].fillna("").tolist()
        t_texts = t_c[text_column].fillna("").tolist()
        
        vectorizer = TfidfVectorizer(analyzer='char_wb', ngram_range=(3, 5), min_df=2, max_df=0.9, dtype=np.float32)
        t_matrix = vectorizer.fit_transform(t_texts)
        s1_matrix = vectorizer.transform(s1_texts)
        
        actual_k = min(top_k, t_matrix.shape[0])
        nn = NearestNeighbors(n_neighbors=actual_k, metric='cosine', n_jobs=-1)
        nn.fit(t_matrix)
        
        s1_ids = s1_c['entity_id'].values
        t_ids = t_c['entity_id'].values
        
        for start in range(0, s1_matrix.shape[0], batch_size):
            end = min(start + batch_size, s1_matrix.shape[0])
            dists, idxs = nn.kneighbors(s1_matrix[start:end])
            
            for i in range(end - start):
                for j in range(actual_k):
                    if dists[i, j] < distance_threshold:
                        all_pairs.append({
                            'entity_id_s1': s1_ids[start + i],
                            'entity_id_cand': t_ids[idxs[i, j]]
                        })
        
        del s1_matrix, t_matrix, vectorizer, nn
        gc.collect()
    
    return pd.DataFrame(all_pairs).drop_duplicates() if all_pairs else pd.DataFrame(columns=['entity_id_s1', 'entity_id_cand'])
```

### Block 5: TF-IDF Address Blocking (catches transliterated names)

```python
def block_tfidf_address(s1, s_target, top_k=10, batch_size=10000, distance_threshold=0.7):
    """TF-IDF on addresses PER COUNTRY. Critical for catching transliteration cases
    where names share zero character overlap but addresses overlap significantly."""
    return block_tfidf(s1, s_target, text_column='norm_address', top_k=top_k, 
                       batch_size=batch_size, distance_threshold=distance_threshold)
```

### Blocking Pipeline: Generate All Candidates

```python
def generate_all_candidates(s1, s_target, source_label):
    """Union of all blocking strategies for S1 -> one target source."""
    print(f"\n  === Blocking S1 -> {source_label} ===")
    
    # Block 1: Name prefix
    print(f"  [1/5] Name Prefix Blocking...")
    b1 = block_name_prefix(s1, s_target)
    print(f"         -> {len(b1)} pairs")
    
    # Block 2: Name token overlap
    print(f"  [2/5] Name Token Blocking...")
    b2 = block_name_tokens(s1, s_target)
    print(f"         -> {len(b2)} pairs")
    
    # Block 3: Address token overlap
    print(f"  [3/5] Address Token Blocking...")
    b3 = block_address_tokens(s1, s_target)
    print(f"         -> {len(b3)} pairs")
    
    # Block 4: TF-IDF on names (per country)
    print(f"  [4/5] TF-IDF Name Blocking (per country)...")
    b4 = block_tfidf(s1, s_target, text_column='norm_name_no_suffix', top_k=10)
    print(f"         -> {len(b4)} pairs")
    
    # Block 5: TF-IDF on addresses (per country)
    print(f"  [5/5] TF-IDF Address Blocking (per country)...")
    b5 = block_tfidf_address(s1, s_target, top_k=10)
    print(f"         -> {len(b5)} pairs")
    
    # UNION all blocks
    all_cands = pd.concat([b1, b2, b3, b4, b5]).drop_duplicates()
    print(f"  UNION total unique pairs: {len(all_cands)}")
    
    del b1, b2, b3, b4, b5
    gc.collect()
    
    return all_cands
```

```python
def run_blocking_pipeline(is_test=False, sample_size=None):
    prefix = "test" if is_test else "train"
    print(f"\n{'='*60}")
    print(f"  BLOCKING PIPELINE \u2014 {prefix.upper()} DATA")
    print(f"{'='*60}")
    if sample_size:
        print(f"  TEST MODE: Only loading first {sample_size} rows!")
    
    print("Loading normalized data...")
    s1_norm = pd.read_csv(os.path.join(NORM_DIR, f"{prefix}_source1.tsv"), sep="\t", nrows=sample_size)
    s2_norm = pd.read_csv(os.path.join(NORM_DIR, f"{prefix}_source2.tsv"), sep="\t", nrows=sample_size)
    s3_norm = pd.read_csv(os.path.join(NORM_DIR, f"{prefix}_source3.tsv"), sep="\t", nrows=sample_size)
    
    # S1 -> S2
    s2_candidates = generate_all_candidates(s1_norm, s2_norm, "Source 2")
    del s2_norm; gc.collect()
    
    # S1 -> S3
    s3_candidates = generate_all_candidates(s1_norm, s3_norm, "Source 3")
    del s3_norm; gc.collect()
    
    # Union & format
    print("\nUnioning S2 + S3 candidates and formatting output...")
    final_candidates = pd.concat([s2_candidates, s3_candidates]).drop_duplicates()
    del s2_candidates, s3_candidates; gc.collect()
    
    grouped = final_candidates.groupby('entity_id_s1')['entity_id_cand'].apply(
        lambda x: ','.join(sorted(set(x)))
    ).reset_index()
    grouped.rename(columns={'entity_id_s1': 'source1_entity_id', 'entity_id_cand': 'candidate_entity_ids'}, inplace=True)
    
    all_s1_ids = pd.DataFrame({'source1_entity_id': s1_norm['entity_id']})
    final_output = pd.merge(all_s1_ids, grouped, on='source1_entity_id', how='left')
    final_output['candidate_entity_ids'] = final_output['candidate_entity_ids'].fillna("")
    
    suffix = "_TEST" if sample_size else ""
    out_file = os.path.join(CANDIDATES_DIR, f"{prefix}_candidate_pairs{suffix}.tsv")
    final_output.to_csv(out_file, sep="\t", index=False)
    print(f"\nSUCCESS! Candidate pairs saved to {out_file}")
    print(f"Total S1 entities: {len(final_output)}")
    print(f"S1 entities with candidates: {(final_output['candidate_entity_ids'] != '').sum()}")
    print(f"S1 singletons (no candidates): {(final_output['candidate_entity_ids'] == '').sum()}")
```

### Run Blocking Pipeline (sample for quick test)

```python
run_blocking_pipeline(is_test=False, sample_size=50000)
```

### Recall Ceiling Evaluation

```python
def evaluate_recall(sample_size=50000):
    """Measure recall ceiling against ground truth."""
    print("\n" + "="*60)
    print("  RECALL CEILING EVALUATION")
    print("="*60)
    
    gt = pd.read_csv(os.path.join(GT_DIR, 'train_ground_truth.tsv'), sep='\t')
    cands = pd.read_csv(os.path.join(CANDIDATES_DIR, 'train_candidate_pairs_TEST.tsv'), sep='\t')
    
    s1_norm = pd.read_csv(os.path.join(NORM_DIR, 'train_source1.tsv'), sep='\t', nrows=sample_size)
    s2_norm = pd.read_csv(os.path.join(NORM_DIR, 'train_source2.tsv'), sep='\t', nrows=sample_size)
    s3_norm = pd.read_csv(os.path.join(NORM_DIR, 'train_source3.tsv'), sep='\t', nrows=sample_size)
    
    s1_ids = set(s1_norm['entity_id'].values)
    valid_targets = set(s2_norm['entity_id'].values) | set(s3_norm['entity_id'].values)
    
    gt_sample = gt[gt['source1_entity_id'].isin(s1_ids)].copy()
    
    total = 0
    found = 0
    missed = []
    
    for _, row in gt_sample.iterrows():
        s1_id = row['source1_entity_id']
        if pd.isna(row['matched_entity_ids']) or str(row['matched_entity_ids']).strip() == '':
            continue
        true_matches = set(str(row['matched_entity_ids']).split(','))
        true_in_sample = true_matches & valid_targets
        if not true_in_sample:
            continue
        
        cand_row = cands[cands['source1_entity_id'] == s1_id]
        if len(cand_row) == 0 or pd.isna(cand_row.iloc[0]['candidate_entity_ids']) or str(cand_row.iloc[0]['candidate_entity_ids']).strip() == '':
            our_cands = set()
        else:
            our_cands = set(str(cand_row.iloc[0]['candidate_entity_ids']).split(','))
        
        for tm in true_in_sample:
            total += 1
            if tm in our_cands:
                found += 1
            else:
                missed.append((s1_id, tm))
    
    print(f"Total true matches (within sample): {total}")
    print(f"Found by blocking: {found}")
    print(f"Missed by blocking: {total - found}")
    if total > 0:
        print(f"RECALL CEILING: {found / total * 100:.2f}%")

evaluate_recall(sample_size=50000)
```
