### Detailed Analysis of Sources

```python
def analyze_source(df, source_name):
    print(f"\n" + "="*50)
    print(f"ANALYSIS FOR {source_name.upper()}")
    print("="*50)
    
    print(f"Shape of dataset: {df.shape} (Rows, Columns)")
    
    unique_ids = df['entity_id'].nunique()
    print(f"Unique entity IDs: {unique_ids}")
    if unique_ids != len(df):
        print(f"WARNING: There are {len(df) - unique_ids} duplicate entity_ids!")
        
    print("\n--- Missing Values ---")
    print(df.isna().sum())
    
    print("\n--- Country Distribution ---")
    if 'country' in df.columns:
        print(df['country'].value_counts(dropna=False))
        
    print("\n--- String Lengths (Characters) ---")
    if 'business_name' in df.columns:
        name_len = df['business_name'].astype(str).apply(len)
        print(f"Business Name - Min: {name_len.min()}, Mean: {name_len.mean():.2f}, Max: {name_len.max()}")
        
    if 'business_address' in df.columns:
        addr_len = df['business_address'].astype(str).apply(len)
        print(f"Business Address - Min: {addr_len.min()}, Mean: {addr_len.mean():.2f}, Max: {addr_len.max()}")
        
    print("\n--- Exact Duplicates ---")
    if 'business_name' in df.columns:
        duplicate_names = df.duplicated(subset=['business_name']).sum()
        print(f"Duplicate Business Names: {duplicate_names}")
        
    if 'business_address' in df.columns:
        duplicate_addresses = df.duplicated(subset=['business_address']).sum()
        print(f"Duplicate Business Addresses: {duplicate_addresses}")
```

```python
analyze_source(s1, "Training Source 1")
analyze_source(s2, "Training Source 2")
analyze_source(s3, "Training Source 3")
```

### Ground Truth Analysis

```python
print(f"\n" + "="*50)
print("ANALYSIS FOR GROUND TRUTH")
print("="*50)
print(f"Shape of Ground Truth: {ground_truth.shape}")

ground_truth['match_count'] = ground_truth['matched_entity_ids'].apply(
    lambda x: 0 if pd.isna(x) or str(x).strip() == "" else len(str(x).split(','))
)

print("\n--- Match Distribution (How many matches per S1 entity) ---")
print(ground_truth['match_count'].value_counts().sort_index())

singletons = ground_truth['match_count'] == 0
print(f"\nTotal Singletons (0 matches): {singletons.sum()} ({(singletons.sum() / len(ground_truth)) * 100:.2f}%)")
```
